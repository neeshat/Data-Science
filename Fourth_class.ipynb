# Importing All Necessary Libraries


```python
import numpy as np
import pandas as pd
```

# 1. Create a NumPy array with 20 random integers between 1 and 100 and its basic operations

```python
randint_array = np.random.randint(1, 101, size=20)
print("NumPy array with 20 random integers between 1 and 100: \n",randint_array)
print("The sum of all the elements: ", np.sum(randint_array))
print("Maximum element is: ", np.max(randint_array))
print("Minimum element is: ", np.min(randint_array))
```

# 2. Creating a DataFrame with 5 rows and 3 columns of random numbers between 1 and 100 and its basic operations

```python
randnums = np.random.randint(1, 101, size =(5,3))
df_randn = pd.DataFrame(randnums)
print("DataFrame with 5 rows and 3 columns of random numbers between 1 and 100: \n",df_randn)
num_of_cols = df_randn.shape[1]
for i in range(0,num_of_cols):
  print(f"Sum of the {i+1}th column is: ", df_randn[i].sum())

  print(f"Mean of the {i+1}th column is: ", df_randn[i].mean())
  print("\n")
```
