# **Task 1: Basic Matplotlib (3 plots)**

## 1.1 Bar chart of gym membership types
## Use: gym['abonoment_type'].value_counts()

```python
gym = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/datasets/gym_membership.csv')
gym['abonoment_type'].value_counts().plot(kind='bar', color='skyblue', edgecolor='black')
plt.grid(True, linestyle='--', linewidth=0.7, alpha=0.7)
plt.title('Gym Membership Types')
plt.xlabel('Membership Types')
plt.ylabel('No. of Members')
plt.tight_layout()
plt.show()
```

## 1.2 Histogram of time in gym
## Use: plt.hist(gym['avg_time_in_gym'])

```python
plt.hist(gym['avg_time_in_gym'], bins=20, color='lightgreen', edgecolor='black')
plt.title('Avg Time Spent in Gym')
plt.xlabel('Time')
plt.ylabel('Members')
plt.grid(True, linestyle='--', linewidth=0.7, alpha=0.7)
plt.tight_layout()
plt.show()
```

## 1.3 Scatter plot: Age vs visits per week
## Use: plt.scatter(gym['Age'], gym['visit_per_week'])

```python
plt.scatter(gym['Age'], gym['visit_per_week'], color='orange', alpha=0.7)
plt.title('Age vs Visit Per Week')
plt.xlabel('Age')
plt.ylabel('visit_per_week')
plt.grid(True, linestyle='--', linewidth=0.7, alpha=0.7)
plt.tight_layout()
plt.show()
```

# **Task 2: Basic Seaborn (3 plots)**

## 2.1 Box plot: Gender vs Age
## Use: sns.boxplot(data=gym, x='gender', y='Age')

```python
sns.boxplot(data=gym, x='gender', y='Age')
plt.title('Gender vs Age')
plt.grid(True, which='both', linestyle='--', linewidth=0.7, alpha=0.7)
plt.tight_layout()
plt.show()
```

## 2.2 Count plot: Days per week
## Use: sns.countplot(data=gym, x='days_per_week')

```python
plt.figure(figsize=(35,8))
sns.countplot(data=gym, x='days_per_week', color='lightblue', edgecolor='black')
plt.title('Members vs Days Per Week')
plt.xlabel('Days Per Week')
plt.ylabel('No. of Members')
plt.xticks(rotation=45, ha='right', fontsize=8)
plt.tight_layout()
plt.show()
```

## 2.3 Bar plot: Average visits by membership type
## Use: sns.barplot(data=gym, x='abonoment_type', y='visit_per_week')

```python
sns.barplot(data=gym, x='abonoment_type', y='visit_per_week', palette='pastel')
plt.grid(True, linestyle='--', linewidth=0.7, alpha=0.7)
plt.title('Membership vs Visit Per Week')
plt.xlabel('Membership')
plt.ylabel('Visit Per Week')
plt.tight_layout()
plt.show()
```
