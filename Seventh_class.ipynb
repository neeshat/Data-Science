```python
#Converting Yes/No to 1/0 - Using LabelEncoder; Since there are 2 categories


from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
cols_to_convert = ['mainroad','guestroom','basement','hotwaterheating','airconditioning','prefarea']
for col in cols_to_convert:
  df[col] = le.fit_transform(df[col])

# Using One-Hot Encoding on furnishingstatus - Since, there are more than 2 categories

df = pd.get_dummies(df, columns=['furnishingstatus'], drop_first = True)
```

```python
df.head()
```

```python
# Check what's in our data

print("Current mainroad values:\n", df['mainroad'].head())
```

```python
# Split into Train and Test
# TODO: Split - 80% train, 20% test

X = df.drop('price', axis=1)
y = df['price']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=20)
```

```python
#Apply StandardScaler
# Area is huge (1650-13300), Bedrooms is small (1-6)!


print(f"Min Area is: ", X_train['area'].min())
print(f"Max Area is: ", X_train['area'].max())

print(f"Min bedrooms: ", X_train['bedrooms'].min())
print(f"Max bedrooms: ", X_train['bedrooms'].max())
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# After Scaling

X_train_scaled_df = pd.DataFrame(X_train_scaled, columns=X_train.columns)
X_train_scaled_df
```

# **Training and testing the model**

```python
# Model Training

from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train_scaled, y_train)

train_score = model.score(X_train_scaled, y_train)
print(f"Traning accuracy: {train_score:.1%}")

test_score = model.score(X_test_scaled, y_test)
print(f"Testing accuracy: {test_score:.1%}")
```

```python
# Performing 5 fold cross validation

from sklearn.model_selection import cross_val_score
score = cross_val_score(model, X_train_scaled, y_train, cv=5)
for i in range(5):
  print(f"Test {i+1} score: {score[i]:.1%}")
print(f"Accuracy of the model is: {score.mean():.1%}")
```
